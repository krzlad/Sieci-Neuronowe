```python
from sklearn import datasets
import pandas as pd

iris = datasets.load_iris()
dane = pd.DataFrame(iris.data, columns=iris.feature_names)

dane['gatunek'] = iris.target
```

```python
print("Wymiary tabeli: ")
print(dane)

print("\n Podział na klasy: ")
print(dane['gatunek'].value_counts())

print("\n Pierwsze trzy kolumny: ")
print(dane.head(3))
```

```python
cechy = dane.iloc[:, :4]

minimum = cechy.min()
maximum = cechy.max()

dane_znormalizowane = (cechy - minimum) / (maximum - minimum)

print("Dane po normalizacji: ")
print(dane_znormalizowane.head(5))

```

```python

import numpy as np

neurony = np.random.rand(3, 4)

print("Tak wyglądają neurony na starcie: ")
print(neurony)
```

```python

def policz_odleglosc(kwiatek, neurony):

    dystans = np.sqrt(np.sum((kwiatek - neurony)**2))

    return dystans
```

```python

powtorzenia = 100
wspolczynnik_uczenia = 0.1

kwiatki_do_uczenia = dane_znormalizowane.values

print("Start uczenia")

for razy in range(powtorzenia):

    for kwiatek in kwiatki_do_uczenia:
        najmniejsza_odleglosc = float('inf')
        indeks_zwyciezca = -1

        for indeks, neuron in enumerate(neurony):

            odleglosc = policz_odleglosc(kwiatek, neuron)
            if odleglosc < najmniejsza_odleglosc:
                najmniejsza_odleglosc = odleglosc
                indeks_zwyciezca = indeks

        neurony[indeks_zwyciezca] = neurony[indeks_zwyciezca] + wspolczynnik_uczenia * (kwiatek - neurony[indeks_zwyciezca])
```

```python
powtorzenia = 50
poczatkowy_wspolczynnik = 0.1
kwiatki_do_uczenia = dane_znormalizowane.values

for razy in range(powtorzenia):
    # Naprawiony wzór i usunięty minus na początku
    wspolczynnik_uczenia = poczatkowy_wspolczynnik * (1 - (razy / powtorzenia))

    if razy % 10 == 0:
        # Naprawiony dwukropek przy .4f
        print(f"Powtórzenie: {razy}: aktualny wspolczynnik: {wspolczynnik_uczenia:.4f}")

    for kwiatek in kwiatki_do_uczenia:
        najmniejsza_odleglosc = float('inf')
        indeks_zwyciezca = -1

        for indeks, neuron in enumerate(neurony):
            odleglosc = policz_odleglosc(kwiatek, neuron)
            if odleglosc < najmniejsza_odleglosc:
                najmniejsza_odleglosc = odleglosc
                indeks_zwyciezca = indeks

        # Dodane brakujące "y" na samym końcu
        neurony[indeks_zwyciezca] = neurony[indeks_zwyciezca] + wspolczynnik_uczenia * (kwiatek - neurony[indeks_zwyciezca])

# Wyrzucone do lewej strony i poprawiony ukośnik
print("\nKoniec uczenia. Ostateczne wagi 3 neuronów:")
print(neurony)
```

```python
tabelka = np.zeros((3, 3), dtype=int)

print("testowanie sieci :D ")

for numer_kwiatka in range(len(kwiatki_do_uczenia)):
    kwiatek = kwiatki_do_uczenia[numer_kwiatka]

    prawdziwy_gatunek = dane['gatunek'][numer_kwiatka]

    najmniejsza_odleglosc = float('inf')
    indeks_zwyciezca = -1

    for indeks, neuron in enumerate(neurony):
        odleglosc = policz_odleglosc(kwiatek, neuron)
        if odleglosc < najmniejsza_odleglosc:
            najmniejsza_odleglosc = odleglosc
            indeks_zwyciezca = indeks

    tabelka[indeks_zwyciezca][prawdziwy_gatunek] += 1

macierz_podsumowujaca = pd.DataFrame(
    tabelka,
    columns=['Setosa', 'Versicolor', 'Virginica'],
    index=['Neuron 1', 'Neuron 2', 'Neuron 3']
    )

print("\n0to macierz przypisań: ")
print(macierz_podsumowujaca)
```
